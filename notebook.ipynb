```python

import pandas as pd

files = [
    "files/erp_goods_receipts.csv",
    "files/invoice_tool_export.csv",
    "files/labels_A.csv",
    "files/taxonomy_A.csv"
]

for file in files:
    print("\n" + "=" * 70)
    print("FILE:", file)

    df = pd.read_csv(file, low_memory=False)

    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 3 rows:")
    print(df.head(3).to_string(index=False))

    print("\nData types:")
    print(df.dtypes.to_string())
```
